## Wykrywanie mrugnięć

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import aseegg as ag
```

```python
df = pd.read_csv("data2.csv")
```

```python
df.columns
```

```python
t = np.linspace(0, len(df['signal'])/200, len(df['signal']))
plt.plot(t, df['signal'])
```

```python
sygnal = df['signal'][200:]
```

```python
t = np.linspace(0, len(sygnal)/200, len(sygnal))
plt.plot(t, sygnal)
plt.xlim([10,15]) #mrugniecia z 10 sek
#kreski to mrugniecia
```

```python
for i in range(200, len(sygnal)): #indeksy, iteracja od 200 probki bo tak jest wczesniej lol (df sygnal i wgl)
    if sygnal[i] < -40000 and sygnal[i-1]> -40000: #probka pozniejsza i wczesniejsza
        print ("blink") #wyszukuje mrugniecia
```

```python
f_axis = np.linspace(0,200,len(df['signal']))
ffted = ag.FFT(df['signal'])
```

```python
plt.plot(f_axis, ffted)
plt.xlim([0,90])
```

# Pobranie danych na sucho

### Instrukcja zbierania danych z OpenBCI Ganglion

1. Zainstaluj pakiety:
>pip install brainflow
2. Podłącz dongle do USB (ten z zestawu Ganglion).
3. Sprawdź port COM:
    - Otwórz Menedżer urządzeń → Porty (COM i LPT)
    - Odczytaj numer (np. COM7, COM9).
    - Ustaw port COM w kodzie (edytuj plik data_collector.py):
    - COM_PORT = 'COM9'   # wpisz swój numer
4. Załóż elektrody:
    - CH1 → czoło
    - REF → ucho
    - D_G → drugie ucho
5. Podłącz baterie i włącz płytkę Ganglion.
6. Uruchom program w konsoli:
>python data_collector.py
7. Zakończ nagrywanie:
    - Ctrl + C
    -(plik CSV zapisze się automatycznie)
